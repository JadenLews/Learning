```python
import pandas as pd
import matplotlib.pyplot as plt

def get_info(history_path):
    name = history_path[-18:-9]
    df = pd.read_csv(history_path)

    print(df.columns)

    # -----------------------------
    # Validation supervised MSE
    # -----------------------------
    plt.figure(figsize=(8,5))
    plt.plot(df["epoch"], df["train_supervised_mse"], label="Train Supervised MSE")
    plt.plot(df["epoch"], df["val_supervised_mse"], label="Val Supervised MSE")
    plt.xlabel("Epoch")
    plt.ylabel("MSE")
    plt.title(f"Supervised MSE {name}")
    plt.legend()
    plt.grid(True)
    plt.show()

    # -----------------------------
    # Darcy loss
    # -----------------------------
    plt.figure(figsize=(8,5))
    plt.plot(df["epoch"], df["train_darcy"], label="Train Darcy")
    plt.plot(df["epoch"], df["val_darcy"], label="Val Darcy")
    plt.xlabel("Epoch")
    plt.ylabel("Darcy Loss")
    plt.title(f"Darcy Loss {name}")
    plt.legend()
    plt.grid(True)
    plt.show()

    # -----------------------------
    # Mask vs Nonmask
    # -----------------------------
    plt.figure(figsize=(8,5))
    plt.plot(df["epoch"], df["val_mask_mse"], label="Val Mask MSE")
    plt.plot(df["epoch"], df["val_nonmask_mse"], label="Val Nonmask MSE")
    plt.xlabel("Epoch")
    plt.ylabel("MSE")
    plt.title(f"Mask vs Nonmask {name}")
    plt.legend()
    plt.grid(True)
    plt.show()

    # -----------------------------
    # Full-image MSE
    # -----------------------------
    plt.figure(figsize=(8,5))
    plt.plot(df["epoch"], df["train_total_mse"], label="Train Total MSE")
    plt.plot(df["epoch"], df["val_total_mse"], label="Val Total MSE")
    plt.xlabel("Epoch")
    plt.ylabel("MSE")
    plt.title(f"Total MSE {name}")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
# CHANGE THIS TO THE RUN YOU WANT
history_path = (
    "recent_analysis/physics_tests/fixed/"
    "fixed_physics_limited_splitnet_attn_darcy_0p0_history.csv"
)
get_info(history_path)
```
