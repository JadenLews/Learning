```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

def get_model_name(history_path):
    # Uses filename without "_history.csv"
    name = Path(history_path).stem
    return name.replace("_history", "")

def compare_histories(path1, path2, label1=None, label2=None):
    df1 = pd.read_csv(path1)
    df2 = pd.read_csv(path2)

    name1 = label1 if label1 else get_model_name(path1)
    name2 = label2 if label2 else get_model_name(path2)

    print("Model 1:", name1)
    print("Columns:", df1.columns.tolist())
    print()
    print("Model 2:", name2)
    print("Columns:", df2.columns.tolist())

    # -----------------------------
    # Supervised MSE
    # -----------------------------
    plt.figure(figsize=(9,5))
    plt.plot(df1["epoch"], df1["train_supervised_mse"], label=f"{name1} Train Supervised")
    plt.plot(df1["epoch"], df1["val_supervised_mse"], label=f"{name1} Val Supervised")
    plt.plot(df2["epoch"], df2["train_supervised_mse"], "--", label=f"{name2} Train Supervised")
    plt.plot(df2["epoch"], df2["val_supervised_mse"], "--", label=f"{name2} Val Supervised")
    plt.xlabel("Epoch")
    plt.ylabel("MSE")
    plt.title("Supervised MSE Comparison")
    plt.legend()
    plt.grid(True)
    plt.show()

    # -----------------------------
    # Darcy loss
    # -----------------------------
    plt.figure(figsize=(9,5))
    plt.plot(df1["epoch"], df1["train_darcy"], label=f"{name1} Train Darcy")
    plt.plot(df1["epoch"], df1["val_darcy"], label=f"{name1} Val Darcy")
    plt.plot(df2["epoch"], df2["train_darcy"], "--", label=f"{name2} Train Darcy")
    plt.plot(df2["epoch"], df2["val_darcy"], "--", label=f"{name2} Val Darcy")
    plt.xlabel("Epoch")
    plt.ylabel("Darcy Loss")
    plt.title("Darcy Loss Comparison")
    plt.legend()
    plt.grid(True)
    plt.show()

    # -----------------------------
    # Mask vs Nonmask
    # -----------------------------
    plt.figure(figsize=(9,5))
    plt.plot(df1["epoch"], df1["val_mask_mse"], label=f"{name1} Val Mask")
    plt.plot(df1["epoch"], df1["val_nonmask_mse"], label=f"{name1} Val Nonmask")
    plt.plot(df2["epoch"], df2["val_mask_mse"], "--", label=f"{name2} Val Mask")
    plt.plot(df2["epoch"], df2["val_nonmask_mse"], "--", label=f"{name2} Val Nonmask")
    plt.xlabel("Epoch")
    plt.ylabel("MSE")
    plt.title("Mask vs Nonmask Comparison")
    plt.legend()
    plt.grid(True)
    plt.show()

    # -----------------------------
    # Full-image MSE
    # -----------------------------
    plt.figure(figsize=(9,5))
    plt.plot(df1["epoch"], df1["train_total_mse"], label=f"{name1} Train Total")
    plt.plot(df1["epoch"], df1["val_total_mse"], label=f"{name1} Val Total")
    plt.plot(df2["epoch"], df2["train_total_mse"], "--", label=f"{name2} Train Total")
    plt.plot(df2["epoch"], df2["val_total_mse"], "--", label=f"{name2} Val Total")
    plt.xlabel("Epoch")
    plt.ylabel("MSE")
    plt.title("Total MSE Comparison")
    plt.legend()
    plt.grid(True)
    plt.show()
```
